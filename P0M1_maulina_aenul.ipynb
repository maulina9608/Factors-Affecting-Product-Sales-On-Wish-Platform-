# DATA CLEANING

```python
df = df[df['price'] <= df['retail_price']] # data yang diambil hanya yang nilai price lebih kecil atau sama dengan nilai retail_price karena data scraping diambil sewaktu COVID-19 oleh karena itu beberapa merchant menaikkan harga
df = df[['title_orig', 'units_sold','price', 'retail_price', 'uses_ad_boosts', 'rating', 'shipping_option_name','shipping_option_price', 'shipping_is_express', 'merchant_rating', 'merchant_id']]
df
```

Data yang diambil adalah data dengan variabel yang bisa menentukan suatu produk bisa terjual lebih banyak atau tidak, yaitu:
- title_orig: nama produk
- price: harga setelah diskon
- retail_price: harga dari merchant
- uses_ad_boosts: penggunaan fitur add boost pada produk, boolean value, 0 adalah tidak, dan 1 ada ya
- rating: rating produk
- shipping_option_name: nama ekspedisi pengiriman
- shipping_option_price: harga ekspsdisi pengiriman
- shipping_is_express: harusnya boolean value, dengan 0 berati tidak dan 1 berati ya
- merchant_rating: rating toko
- merhcant_id: id toko
- units_sold: jumlah produk yang terjual

Tidak ada data NUll, dan tipe data sudah sesuai kecuali 'uses_ad_boosts' dan 'shipping_is_express'. 
pengubahan nama kolom: 'title_orig' menjadi 'product_name' ; 'price' menjadi 'discounted_price' ; dan 'rating' menjadi 'product_rating'

```python
# penggantian tipe data
df['uses_ad_boosts'] = df['uses_ad_boosts'].astype(bool)
df['shipping_is_express'] = df['shipping_is_express'].astype(bool)
# penggantian nama kolom
df = df.rename(columns={
    'title_orig': 'product_name',
    'price': 'discounted_price',
    'rating': 'product_rating'
})
```

```python
# cek dan hapus duplicate
df.duplicated().sum()
df = df.drop_duplicates()
```

```python
# membuat produk id
df['Product_ID'] = ['P{:04d}'.format(i+1) for i in range(len(df))]
cols = ['Product_ID'] + [col for col in df.columns if col != 'Product_ID']
df = df[cols]
```

Pembuatan Product_ID dilakukan untuk menggantikan composite key (product_name, merchant_id) dengan primary key baru yang bersifat unik. 

```python
# membuat kolom baru: 'retail_price'-'discounted_price'
df['price_difference'] = df['retail_price'] - df['discounted_price']
```

## Descriptive Statistics

```python
# Rata-rata
print("Rata-rata:")
print(df_num.mean()) 

# Median
print("\nMedian:")
print(df_num.median()) 

# Standar deviasi
print("\nStandar Deviasi:") 
print(df_num.std()) # menunjukkan seberapa besar variasi nilai data

# Skewness
print("\nSkewness:") 
print(df_num.skew()) # menunjukkan arah kemiringan distribusi data

# Kurtosis
print("\nKurtosis:")
print(df_num.kurt()) # menunjukkan besarnya outlier data
```

Melihat nilai rata-rata unit_sold dan price_difference terlampau jauh lebih besar dari median, bisa kita lihat bahwa kemungkinan besar data miring ke kanan. Persebaran data juga menunjukkan angka yang jauh dari kategori data normal di variabel tersebut. Begitu pula dengan skewness dan kurtosis.

Data yang lain memang tidak terlalu ideal penyebaran dan kemiringannya tetapi hal tersebut kita abaikan karena tidak dianggap memengaruhi analisis secara ekstrim. 

### Menghapus Outlier dengan Turkey's Rule

```python
def remove_outliers(df_num, column): # buat fungsi agar tidak nulis kode dua kali
    Q1 = df_num[column].quantile(0.25)
    Q3 = df_num[column].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    return df_num[(df_num[column] >= lower) & (df_num[column] <= upper)]


df_clean = remove_outliers(df_num, 'units_sold')
df_clean = remove_outliers(df_clean, 'price_difference')
```

```python
# Rata-rata
print("Rata-rata:")
print(df_clean.mean()) 

# Median
print("\nMedian:")
print(df_clean.median()) 

# Skewness
print("\nSkewness:") 
print(df_clean.skew()) # menunjukkan arah kemiringan distribusi data
```

```python
# menambah kolom lain yang belum masuk ke data frame yang bersih
df_clean['uses_ad_boosts'] = df.loc[df_clean.index, 'uses_ad_boosts']
df_clean['shipping_is_express'] = df.loc[df_clean.index, 'shipping_is_express']
df_clean['product_name'] = df.loc[df_clean.index, 'product_name']
df_clean['Product_ID'] = df.loc[df_clean.index, 'Product_ID']
```

```python
df_clean.to_csv('cleaned_data.csv', index=False)
```

## Inferential Statistics

### Uji Korelasi

```python
# uji korelasi units_sold vs price_difference menggunakan Spearman Rank Correlation
import scipy.stats as stats
r, p = stats.spearmanr(df_clean['units_sold'], df_clean['price_difference'])
print(f'units_sold vs price_difference (r, p): {r},{p}')

# uji korelasi units_sold vs uses_ad_boosts menggunakan Korelasi Point-Biserial
r, p = stats.pointbiserialr(df_clean['uses_ad_boosts'], df_clean['units_sold'])
print(f'units_sold vs uses_ad_boosts (r, p): {r},{p}')

# uji korelasi units_sold vs product_rating menggunakan Spearman Rank Correlation
r, p = stats.spearmanr(df_clean['units_sold'], df_clean['product_rating'])
print(f'units_sold vs product_rating (r, p): {r},{p}')

# units_sold vs shipping_option_price menggunakan Spearman Rank Correlation
r, p = stats.spearmanr(df_clean['units_sold'], df_clean['shipping_option_price'])
print(f'units_sold vs shipping_option_price (r, p): {r},{p}')

# units_sold vs shipping_is_express menggunakan Korelasi Point-Biserial
r, p = stats.pointbiserialr(df_clean['shipping_is_express'], df_clean['units_sold'])
print(f'units_sold vs shipping_is_express (r, p): {r},{p}')

# units_sold vs merchant_rating menggunakan Spearman Rank Correlation
r, p = stats.spearmanr(df_clean['units_sold'], df_clean['merchant_rating'])
print(f'units_sold vs merchant_rating (r, p): {r},{p}')

# units_sold vs 'discounted_price' menggunakan Spearman Rank Correlation
r, p = stats.spearmanr(df_clean['units_sold'], df_clean['discounted_price'])
print(f'units_sold vs discounted_price (r, p): {r},{p}')
```

Dari uji korelas data di atas dapat kita simpulkan sementara bahwa:

1. units_sold vs price_difference

Koefisien korelasi Spearman (r: -0.061, p: 0.092) menunjukkan korelasi negatif yang sangat lemah antara jumlah produk terjual dan selisih harga. Meskipun angka ini menyiratkan bahwa selisih harga yang lebih tinggi mungkin sedikit terkait dengan penjualan yang lebih rendah, hubungan ini tidak signifikan secara statistik. Oleh karena itu, kita tidak dapat menyimpulkan bahwa price_difference memiliki pengaruh yang berarti terhadap units_sold.

2. units_sold vs uses_ad_boosts

Korelasi Point-Biserial (r: -0.041, p: 0.25) menunjukkan korelasi negatif yang sangat lemah, dan nilai p menunjukkan angka yang jauh dari nilai standard. Oleh karena itu, kita tidak dapat menyimpulkan bahwa add boosts memiliki pengaruh yang berarti terhadap unit_sold. 

3. units_sold vs product_rating

Koefisien korelasi Spearman (r: -0.013, p: 0.72) menunjukkan korelasi negatif yang sangat lemah antara jumlah produk terjual dan product_rating. nilai p menunjukkan angka yang jauh dari nilai standard. Oleh karena itu, kita tidak dapat menyimpulkan bahwa product_rating memiliki pengaruh yang berarti terhadap unit_sold. 

4. units_sold vs shipping_option_price
Koefisien korelasi Spearman (r: 0.06, p: 0.072) menunjukkan korelasi positif dengan angka yang sangat kecil. oleh karena itu kita tidak bisa menyimpulkan bahwa product_rating memiliki pengaruh yang berarti terhadap harga shipping.

5. units_sold vs shipping_is_express
Korelasi Point-Biserial (r: 0.087, p: 0.015) menunjukkan bahwa uji korelasi ini secara statistik menunjukkan korelasi positif meski sangat lemah, karena nilai p-nya <0.05. Oleh karena itu, pengiriman ekspress mengindikasikan ada kenaikan jumlah produk terjual meski sangat kecil korelasinya. 

6. units_sold vs merchant_rating
Koefisien korelasi Spearman (r: 0.18, p: 2.22) menunjukkan korelasi positif yang sangat lemah antara jumlah produk terjual dan merchant_rating. hal ini menandakan bahwa merchant dengan rating yang tinggi memiliki kecenderungan untuk menjual produk lebih banyak meski tidak signifikan.  

7. units_sold vs discounted_price
Koefisien korelasi Spearman (r: 0.05, p: 1.13) menunjukkan korelasi positif yang sangat lemah antara jumlah produk terjual dan harga setelah diskon. hal ini menandakan bahwa produk dengan harga yang semakin tinggi memiliki kecenderungan untuk menjual produk lebih banyak meski tidak signifikan. 


### Visualization
